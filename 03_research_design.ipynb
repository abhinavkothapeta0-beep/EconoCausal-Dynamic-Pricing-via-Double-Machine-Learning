# Research Design – EconoCausal

## 1. Research Question

How do experimental electricity price signals affect household electricity consumption?

## 2. Study Context

The iFlex field experiment was conducted in several regions of Norway over two winter periods from early 2020 to spring 2021.

## 3. Study Information

- Participating households: 3,746 (reported in the related experiment analysis).
- Price signals tested: 14.
- Consumption measurement: Hourly.
- Price notification: One day ahead.
- Additional information: Household surveys and temperature data.

## 4. Proposed Research Variables

### Treatment Variable
Exposure or assignment to an experimental electricity price signal.

### Outcome Variable
Household electricity consumption, measured in kWh.

### Potential Control Variables
- Outdoor temperature.
- Hour of day.
- Day of week.
- Household characteristics.
- Experiment phase.

These variables will be confirmed after examining the data and experiment design.

## 5. Research Design Questions

1. What was the exact treatment-assignment mechanism?
2. What was the unit of randomization?
3. How were control groups selected?
4. How frequently did households receive price signals?
5. Were all 14 signals available to every treatment group?
6. Were there differences between experiment phases?

## 6. Proposed Analysis

- Descriptive statistical analysis.
- Exploratory data analysis.
- Treatment and comparison group analysis.
- Causal effect estimation, if identification assumptions are supported.
- Double Machine Learning, if an appropriate estimator can be justified.

## 7. Current Status

The research question and initial study context are documented.

The exact treatment definition, assignment mechanism and causal estimation strategy remain under investigation.

## 8. Treatment and Control Group Structure

### 8.1 Control Groups

Control households did not receive the experimental electricity price signals.

Their electricity consumption provides a comparison for understanding consumption patterns without the experimental price intervention.

### 8.2 Treatment Groups

Treatment households received experimental electricity price signals on selected experiment days.

They were also provided with information about possible ways to adjust electricity consumption during high-price periods.

### 8.3 Treatment Definition

The proposed treatment is assignment or exposure to an experimental electricity price signal.

The actual treatment variable will be finalized after inspecting the participant and hourly consumption records.

### 8.4 Outcome Definition

The primary outcome is hourly household electricity consumption, measured in kWh.

### 8.5 Important Considerations

- Treatment assignment must be distinguished from actual household response.
- Different treatment groups received different price signals.
- Experiment days varied across groups.
- Control-group observations must be identified correctly.
- Repeated observations from the same household must be considered.
- Differences between Phase 1 and Phase 2 must be investigated.

### 8.6 Current Research Decision

A causal analysis appears promising because the experiment used randomized treatment and control groups.

However, the exact assignment mechanism, treatment timing and comparison strategy must be verified before selecting a causal estimator.

## 9. Research Variable Definition

### 9.1 Treatment Variable

The treatment variable represents whether a household was assigned to receive an experimental electricity price signal.

The specific treatment definition will depend on the experiment assignment records and timing.

### 9.2 Outcome Variable

The primary outcome is hourly household electricity consumption, measured in kilowatt-hours (kWh).

### 9.3 Potential Control Variables

The following variables may help account for other influences on electricity consumption:

- Outdoor temperature.
- Hour of day.
- Day of week.
- Household characteristics.
- Experiment phase.

### 9.4 Proposed Causal Relationship

The research investigates whether experimental electricity price signals influence household electricity consumption.

### 9.5 Validation Requirements

Before modelling, we must verify:

1. How treatment assignment is recorded.
2. How consumption is measured.
3. Whether observations are available for both treatment and control households.
4. Whether household identifiers are available.
5. Whether temperature and household characteristics are available.
6. Whether the experiment phases can be analyzed separately.

### 9.6 Current Status

Initial research variables have been proposed. Their final definitions will be confirmed after examining the original dataset.
